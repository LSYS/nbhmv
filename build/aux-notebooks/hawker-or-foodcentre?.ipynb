```python
# huh NEA?
import os
import geopandas as gpd
import pandas as pd
pd.set_option('display.max_columns', 100)

from utilities import get_html_table_metadata

import warnings
warnings.filterwarnings('ignore')
```

```python
FP_HAWKERS = os.path.join("raw-data", "HawkerCentresGEOJSON.geojson")

gdf_hawkers = (
    gpd.read_file(FP_HAWKERS)
    .pipe(get_html_table_metadata)
)
hawkers1_name = gdf_hawkers["NAME"]
hawkers1_pc = gdf_hawkers["ADDRESSPOSTALCODE"]
gdf_hawkers
```

```python
FP_HAWKERS2 = os.path.join("raw-data", "NEAMarketandFoodCentre.geojson")
gdf_hawkers2 = (
    gpd.read_file(FP_HAWKERS2)
    .pipe(get_html_table_metadata)
)
hawkers2_name = gdf_hawkers2["NAME_OF_CENTRE"]
hawkers2_pc = gdf_hawkers2["POSTAL_CODE"]
gdf_hawkers2
```

```python
hawkers1_name = [x.lower() for x in hawkers1_name]
hawkers2_name = [x.lower() for x in hawkers2_name]
overlap = list(set(hawkers1_name) & set(hawkers2_name))
print(overlap)
print(len(overlap))
```

```python
ix = 0
for h in hawkers1_name:
    if h in hawkers2_name:
        print(h)
        ix +=1
ix        
```

```python
ix = 0
for h in hawkers2_name:
    if h in hawkers1_name:
        print(h)
        ix += 1
ix        
```

### postal code overlap

```python
hawkers1_pc_clean = [str(x).strip().lower() for x in hawkers1_pc]
hawkers2_pc_clean = [str(x).strip().lower() for x in hawkers2_pc]
overlap = list(set(hawkers1_pc) & set(hawkers2_pc))
print(overlap)
print(len(overlap))
```

```python
ix = 0
for h in hawkers1_pc_clean:
    if h not in hawkers2_pc_clean:
        print(h)
        ix += 1
print(ix)
```

```python
ix = 0
for h in hawkers2_pc_clean:
    if h not in hawkers1_pc_clean:
        print(h)
        ix += 1
print(ix)
```
